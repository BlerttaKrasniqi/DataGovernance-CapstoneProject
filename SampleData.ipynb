```python

metadata_pdf = pd.read_csv(metadata_url)
customers_pdf = pd.read_csv(customers_url)
orders_pdf = pd.read_csv(orders_url)
products_pdf = pd.read_csv(products_url)
payments_pdf = pd.read_csv(payments_url)
billing_pdf = pd.read_csv(billing_url)
```

```python
# Convert Pandas DataFrames to Spark DataFrames
metadata_df = spark.createDataFrame(metadata_pdf)
customers_df = spark.createDataFrame(customers_pdf)
orders_df = spark.createDataFrame(orders_pdf)
products_df = spark.createDataFrame(products_pdf)
payments_df = spark.createDataFrame(payments_pdf)
billing_df = spark.createDataFrame(billing_pdf)
```

```python
# Save as Bronze tables
metadata_df.write.mode("overwrite").saveAsTable("bronze_metadata")

customers_df.write.mode("overwrite").saveAsTable("bronze_customers")
orders_df.write.mode("overwrite").saveAsTable("bronze_orders")
products_df.write.mode("overwrite").saveAsTable("bronze_products")
payments_df.write.mode("overwrite").saveAsTable("bronze_payments")
billing_df.write.mode("overwrite").saveAsTable("bronze_billing_transactions")
```

```python
# Verify tables were created
spark.sql("SHOW TABLES").show()
```
